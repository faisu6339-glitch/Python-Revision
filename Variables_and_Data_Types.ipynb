```python
# 1.Create variables of int, float, str, bool.

a=int(input("Enter a integer number"))
b=float(input("Enter a float number"))
c=str(input("Enter a string"))
d=bool(input("Enter a boolean"))
print(a,b,c,d)
```

```python
# 2.Print the type of each variable.
# Print types
print("\nData Types:")
print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

```python
# 3.Convert string "100" into an integer.
a = "100"

b = int(a)

print(b)
print(type(b))
```

```python
# 4.Convert integer 100 into a float.
a = "100"

b = float(a)

print(b)
print(type(b))
```

```python
# 5.Convert float 25.5 into an integer.
a = 25.5

b = int(a)

print(b)
print(type(b))
```

```python
# 6.Swap two variables.
a = 10
b = 20

print("Before swapping:")
print("a =", a)
print("b =", b)

temp = a
a = b
b = temp

print("After swapping:")
print("a =", a)
print("b =", b)
```

```python
# 7.Swap two variables without using a third variable
a = 10
b = 20

print("Before swapping:")
print("a =", a)
print("b =", b)

a, b = b, a

print("After swapping:")
print("a =", a)
print("b =", b)
```
