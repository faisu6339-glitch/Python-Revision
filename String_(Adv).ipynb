# Question 1:
Write a Python program to reverse a given string without using slicing ([::-1]).

Example:

Input:

Enter a string: programming

```python
s=input("Enter a string:")
reverse=""

for char in s:
  reverse=char+reverse

print("The reverse string is:",reverse)
```

# Question 2: Count Vowels and Consonants
Difficulty: Beginner–Intermediate
Interviewer asks:

Write a Python program to count the total number of vowels and consonants in a given string.

Example:

Input:
Enter a string: Programming


Expected output:
Vowels: 3
Consonants: 8


Conditions:
1. Use a loop.
2. Ignore spaces, digits and special characters.
3. Your program should work with uppercase and lowercase letters.
4. Don't use external libraries.

```python
string=input("Enter a string:")
vowel_list=["a","e","i","o","u"]

vowels=0
consonants=0

for ch in string.lower():
  if ch in vowel_list:
    vowels=vowels+1
  elif ch.isalpha():
    consonants=consonants+1

print("Vowels:",vowels)
print("Consonants:",consonants)
```

# Question 4: Remove Duplicate Characters
Difficulty: Intermediate
Interviewer asks:

Write a Python program to remove duplicate characters from a string while maintaining the original order.

Input:
Enter a string: programming


Expected output:
progamin


Conditions:
1. Use a loop.
2. Don't use set().
3. Maintain the original character order.
4. Don't use a dictionary.

```python
string=input("Enter a string:")
result=""

for char in string:
  if char not in result:
    result+=char
print("String after removing duplicates:", result)
```

# Question 5: Find Duplicate Characters and Their Frequencies
Difficulty: Intermediate
Interviewer asks:

Write a Python program to find all duplicate characters in a string and display their frequencies.

Input:
Enter a string: programming


Expected output:
Duplicate characters:

r: 2
g: 2
m: 2


Conditions:
1. Use nested loops.
2. Don't use Counter.
3. Don't use a dictionary.
4. Don't print duplicate characters more than once.

```python
string = input("Enter a string:")

visited = ""

for char in string:

    if char not in visited:

        count = 0

        for ch in string:
            if char == ch:
                count += 1

        if count > 1:
            print(char, ":", count)

        visited += char
```
