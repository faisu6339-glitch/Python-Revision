# Print all elements using a for loop.

```python
numbers = ()

for i in range(5):
    num = int(input("Enter an integer: "))
    numbers += (num,)

print("Tuple:", numbers)
```

# Print all elements using a while loop.

```python
numbers = (10, 20, 30, 40, 50)

i = 0

while i < len(numbers):
    print(numbers[i])
    i += 1
```

# Check whether an element exists in a tuple.

```python
numbers =(10,20,30,40,50,60)
search=int(input("Enter the elements:"))
found=False

for num in numbers:
  if num == search:
    found=True
    break

if found:
  print("Element found")
else:
  print("Element not found")
```

# Convert a list into a tuple.

```python
numbers = [10, 20, 30, 40, 50]

result = tuple(numbers)

print("Original List:", numbers)
print("Converted Tuple:", result)
print("Data Type:", type(result))
```

# Convert a tuple into a list.

```python
numbers = (10, 20, 30, 40, 50)

result = list(numbers)

print("Original List:", numbers)
print("Converted Tuple:", result)
print("Data Type:", type(result))
```

# Find the maximum and minimum elements.

```python
numbers = (45, 12, 89, 23, 67, 5, 100)

maximum = max(numbers)
minimum = min(numbers)

print("Tuple:", numbers)
print("Maximum element:", maximum)
print("Minimum element:", minimum)
```
