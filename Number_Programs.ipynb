# 1.Check whether a number is prime.

```python
num = int(input("Enter a number: "))

if num <= 1:
    print(num, "is not a prime number")
else:
  for i in range(2 , num):
    if num % i==0:
      print(num,"is not a prime number")
      break
  else:
    print(num,"is a prime number")
```

# 2.Find factors of a number.

```python

num = int(input("Enter a number: "))

for i in range(1, num + 1):
    if num % i == 0:
        print(i)
```

# 3. Find the factorial of a number

```python
num=int(input("Enter a number:"))
fact=1

for i in range(1,num+1):
  fact*=i
print("Factorial of",num,"is",fact)
```

## 4.Print prime numbers between 1 and 100.

```python
for num in range(1,101):
  if num > 1:
    for i in range(2,num):
      if num % i == 0:
        break
    else:
      print(num)
```

# 5.Count prime numbers.

```python

count = 0

for num in range(2, 101):
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        count += 1

print("Total prime numbers:", count)
```

# 6.Check Armstrong number.

```python
num=int(input("Enter a number:"))

temp=num
digits=len(str(num))
rev=0

while temp > 0:
  digit=temp % 10
  rev +=digit**digits
  temp //=10
if rev == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")
```

# 7.Check palindrome number.

```python
num=int(input("Enter a number:"))

temp=num
rev =0

while temp > 0:
  digit=temp % 10
  rev = rev * 10 + digit
  temp //=10

if rev == num:
    print(num, "is a palindrome number")
else:
    print(num, "is not a palindrome number")


```

# 9.Find LCM

```python

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

if a > b:
    greater = a
else:
    greater = b

while True:
    if greater % a == 0 and greater % b == 0:
        print("LCM =", greater)
        break
    greater += 1
```

# 10.Find GCD (Greatest Common Divisor)

```python

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

smallest = min(a, b)
gcd = 1

for i in range(1, smallest + 1):
    if a % i == 0 and b % i == 0:
        gcd = i

print("GCD =", gcd)
```
