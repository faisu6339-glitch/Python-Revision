# Check whether a number is prime.

```python
num = int(input("Enter a number: "))

if num <= 1:
    print(num, "is not a prime number")
else:
  for i in range(2 , num):
    if num % i==0:
      print(num,"is not a prime number")
      break
  else:
    print(num,"is a prime number")
```

# Print prime numbers between 1 and 100.

```python
for num in range(1,101):
  if num > 1:
    for i in range(2,num):
      if num % i == 0:
        break
    else:
      print(num)
```
