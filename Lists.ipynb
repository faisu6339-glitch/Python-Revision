### Introduction to Python Lists

A **list** is one of the most versatile and commonly used data structures in Python.
It is used to store an ordered collection of items, which can be of different data types (integers, strings, floats, or even other lists).

#### Key Characteristics of Lists:
* **Ordered**: The items have a defined order, and that order will not change (unless explicitly modified).
* **Mutable**: You can change, add, and remove items in a list after it has been created.
* **Allow Duplicates**: Lists can have multiple items with the same value.

```python
# 1. Creating a list
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed_list = ["apple", 42, 3.14, True]

print("Fruits list:", fruits)
print("Mixed list:", mixed_list)
```

#### Accessing List Items

List items are **indexed** starting from `0` for the first element. You can also use negative indexing starting from `-1` to access elements from the end of the list.

```python
# 2. Accessing items
print("First fruit:", fruits[0])
print("Last fruit (negative index):", fruits[-1])

# Slicing a list (extracting a sublist)
# Syntx: list[start:end] (end index is excluded)
print("Sliced numbers (first 3):", numbers[0:3])
```

#### Modifying List Items & Methods

Since lists are mutable, we can change individual items, add new items with `.append()` or `.insert()`, and remove items using `.remove()` or `.pop()`.

```python
# 3. Modifying a list
fruits[1] = "blueberry"
print("Modified list:", fruits)

# Adding items
fruits.append("orange")
print("After append:", fruits)

# Removing items
fruits.remove("apple")
print("After removing 'apple':", fruits)

# Finding length
print("List length:", len(fruits))
```

```python
# Find duplicate
names=["Faisal Sheikh","Aaliya Fatma","Shadmani","Ummehani","Shabnam","Zainab","Shabnam","Noor Aamna"]
for name in names:
  if names.count(name)>1:
    print(name,"is duplicate")

# Using a for loop without count()

seen=[]
duplicates=[]
for name in names:
  if name not in seen:
    seen.append(name)
  else:
    duplicates.append(name)
print(duplicates)
```

```python
# Remove duplicates.
names=["Faisal Sheikh","Aaliya Fatma","Shadmani","Ummehani","Shabnam","Zainab","Shabnam","Noor Aamna"]

unique_names = []

for name in names:
    if name not in unique_names:
        unique_names.append(name)

print(unique_names)
```

```python
# Find second largest number.

num1 = [4, 5, 77, 85, 999, 455.1000,988]

largest = second_largest = float('-inf')

for n in num1:
    if n > largest:
        second_largest = largest
        largest = n
    elif n > second_largest and n != largest:
        second_largest = n

print("Largest number:", largest)
print("Second largest number:", second_largest)
```

```python
# second-smallest number

num1 = [4, 5, 77, 85, 999, 455]

smallest = second_smallest = float('inf')

for n in num1:
    if n < smallest:
        second_smallest = smallest
        smallest = n
    elif n < second_smallest and n != smallest:
        second_smallest = n

print("Smallest number:", smallest)
print("Second smallest number:", second_smallest)
```
