#Check whether a year is a leap year.

```python
year=int(input("Enter a year:"))

if year % 400 ==0:
  print(year,"is a leap year")
elif year % 100 ==0:
  print(year,"is not a leap year")
elif year % 4 ==0:
  print(year,"is a leap year")
else:
  print(year,"is not a leap year")
```

# Calculate electricity bill based on units.

```python

unit_consumed = int(input("Enter the units: "))

price_per_unit = 36
additional_charge = 500

if unit_consumed < 300:
    print("Your electricity bill is free")

elif unit_consumed <= 500:
    bill = unit_consumed * price_per_unit
    print("Your electricity bill is ₹", bill)

else:
    bill = unit_consumed * price_per_unit + additional_charge
    print("Your electricity bill is ₹", bill)
```

# Create a restaurant billing program.

```python

menu = {
    "Pizza": 250,
    "Burger": 55,
    "Pasta": 95,
    "Chowmin": 85,
    "Gosht": 950,
    "Nehari": 75,
    "Mutton": 800,
    "Chicken": 200
}

total_bill = 0

while True:
    food = input("Enter food item (or 'done' to finish): ").strip().title()

    if food == "Done":
        break

    if food in menu:
        quantity = int(input("Enter quantity: "))

        price = menu[food]
        item_total = price * quantity

        total_bill += item_total

        print(f"{food}: ₹{price} x {quantity} = ₹{item_total}")

    else:
        print("Sorry, this food item is not available!")

print("\n----- Final Restaurant Bill -----")
print("Total bill: ₹", total_bill)
print("Thank you for visiting!")
```
