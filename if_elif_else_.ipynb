# Check whether a number is a perfect square without using a square-root function.

```python
num = int(input("Enter a number: "))

if num < 0:
    print(num, "is not a perfect square.")
else:
    i = 1
    temp = num
    # Subtract consecutive odd numbers
    while temp > 0:
        temp -= i
        i += 2  # Move to the next odd number (1, 3, 5, 7...)

    # If temp hits 0 exactly, it's a perfect square
    if temp == 0:
        print(num, "is a perfect square")
    else:
        print(num, "is not a perfect square")
```

# Calculate parking charges based on hours parked.

```python
per_hour_charges = 100
additional_charge = 20

hours_parked = int(input("Enter the number of hours parked: "))

if hours_parked <= 2:
    print("Parking is FREE")
elif hours_parked <= 5:
    # Subtract the 2 free hours so they only pay for the extra hours
    cost = (hours_parked - 2) * per_hour_charges
    print("Parking cost is", cost)
else:
    cost = ((hours_parked - 2) * per_hour_charges) + additional_charge
    print("Parking cost is", cost)
```

# Check whether a password satisfies length, digit, uppercase, and special-character requirements.

```python
import string


def validate_password(password):
  min_length = 8
  special_characters = ["@", "#", "$", "&"]

  # Flags to track requirements
  upper_case = False
  has_special = False

  # 1. Check length
  if len(password) < min_length:
    print("Password must be at least 8 characters long.")
    return False

  # 2. Loop through the password to check uppercase and special characters
  for char in password:
    if char.isupper():
      upper_case = True
    if char in special_characters:
      has_special = True

  # 3. Validate uppercase
  if not upper_case:
    print("Password must contain at least one uppercase letter.")
    return False

  # 4. Validate special characters from your list
  if not has_special:
    print(f"Password must contain at least one special character: {special_characters}")
    return False

  print("Success! Password is valid.")
  return True


# Test it out
user_password = input("Enter your password: ")
validate_password(user_password)
```

# Check whether three numbers can be the sides of a right-angled triangle.

```python
side1 = int(input("Enter the length of first side: "))
side2 = int(input("Enter the length of second side: "))
side3 = int(input("Enter the length of third side: "))

# Check if any combination satisfies the Pythagorean theorem (a^2 + b^2 = c^2)
if (
    (side1**2 + side2**2 == side3**2)
    or (side1**2 + side3**2 == side2**2)
    or (side2**2 + side3**2 == side1**2)
):
  print("It is a Right-angled Triangle")
else:
  print("It is not a Right-angled Triangle")
```
