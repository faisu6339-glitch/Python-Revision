# 1.Second-largest distinct number
Beginner–Intermediate
Lists · Conditions · Loops

Interviewer says:

Given a list of integers, write a Python program to find the second-largest distinct number without using sort(), sorted(), or external libraries.

Example test cases
Input	Expected output
[10, 5, 8, 20, 15]	15
[4, 4, 3, 2]	3
[-1, -5, -3]	-3
[7, 7, 7]	No second-largest distinct number

Requirements
- Don't sort the list.
- Handle duplicate values correctly.
- Handle the case where no second-largest distinct number exists.
- Explain your approach after writing the code.

```python
list=[10,205,8,98,455,258,789,52]
largest=float('-inf')
second_largest=float('-inf')

for i in list:
  if i > largest:
    second_largest=largest
    largest=i
  elif i>second_largest and i!=largest:
    second_largest=i
print(second_largest)
```

# Question 2: Digital Root
Beginner–Intermediate

Topics: while loop · Modulo · Integer division

Interviewer asks:

Write a Python program that repeatedly adds the digits of a positive integer until only a single digit remains.

Examples
Input	Working	Output
9875	9 + 8 + 7 + 5 = 29, then 2 + 9 = 11, then 1 + 1 = 2	2
456	4 + 5 + 6 = 15, then 1 + 5 = 6	6
9	Already a single digit	9

Requirements
- Use loops and arithmetic operations.
- Don't convert the number to a string.
- Don't use a built-in digital-root function.
- Print the final single digit.

```python

num = int(input("Enter a positive integer: "))

while num >= 10:
    total = 0

    while num > 0:
        digit = num % 10
        total += digit
        num = num // 10

    num = total

print(num)
```

# Question 3 — Find the Product of Digits at Even Positions
Level: Beginner–Intermediate
Topics: while loop · Modulo (%) · Integer division (//) · Counter · Multiplication
Problem statement
Write a Python program to find the product of digits present at even positions in a positive integer.

Assume positions are counted from left to right, starting from 1.

Example test cases:

Input	  
 Even-position 123456
 digits
 Expected output
	     2, 4, 6	           48
9875	       8, 5	               40
12345	       2, 4	               8
2468	       4, 8	               32
111	          1	                 1

Requirements
1. Use a while loop.
2. Don't convert the number into a string.
3. Don't use lists.
4. Use arithmetic operations only.
5. Find the product of digits at even positions.

```python

num = int(input("Enter a number: "))

product = 1
position = 0
temp = num

# Count the number of digits
while temp > 0:
    position += 1
    temp //= 10

# Extract digits from right to left
while num > 0:
    digit = num % 10

    if position % 2 == 0:
        product *= digit

    position -= 1
    num //= 10

print("Product of even-position digits:", product)
```

# Find the sum of factorials of alternate digits

```python

num = int(input("Enter a number: "))

total = 0
position = 1

while num > 0:
    digit = num % 10

    if position % 2 != 0:
        fact = 1
        i = 1

        while i <= digit:
            fact *= i
            i += 1

        total += fact

    position += 1
    num //= 10

print("Sum of factorials:", total)
```
